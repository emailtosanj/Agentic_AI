```python
import requests
import time
import datetime


# Code added
import os
from dotenv import load_dotenv


# 1. Configuration
API_KEY = "sk-admin-..."  # Must be an Admin Key or have 'organization.read' scope
ORG_ID = "org-..."        # Optional: strictly required if you belong to multiple orgs
URL = "https://api.openai.com/v1/organization/costs"

def get_daily_costs():
    # Calculate start time (Unix timestamp for 30 days ago)
    
    # This API is usually added called by ADMIN to get the costs.
    # API needs admin access.

    print('start 1 get_daily_costs') 

    start_time = int(time.time() - (30 * 24 * 60 * 60))

    API_KEY = os.getenv('OPENAI_API_KEY')
    print('API_KEY ***', API_KEY)
    #this is a service account key
    
    # Used the Admin API key and it did not work.
    # start 1 get_daily_costs
    # Error 403: {"error":"You have insufficient permissions for this operation. Missing scopes: api.usage.read. Check that you have the correct role in your organization, and if you're using a restricted API key, that it has the necessary scopes."}
    # None
    
    headers = {
        "Authorization": f"Bearer {API_KEY}",
        "OpenAI-Organization": "org-ncvHtORQCyyH7hmBTsKpc6U9",
        "Content-Type": "application/json"
    }
    
    # 2. Define Parameters
    params = {
        "start_time": start_time,
        "limit": 30,           # Max results
        "bucket_width": "1d",  # Group by "1d" (day) or "1h" (hour)
        # "group_by": "project_id" # Optional: Split costs by project
    }

    # 3. Make Request
    response = requests.get(URL, headers=headers, params=params)
    
    if response.status_code == 200:
        data = response.json().get("data", [])
        print(f"{'Date':<15} | {'Cost (USD)':<10}")
        print("-" * 30)
        
        for day in data:
            # Convert timestamp to readable date
            # Note: API returns 'start_time' for the bucket
            date_str = datetime.datetime.fromtimestamp(day['start_time']).strftime('%Y-%m-%d')
            amount = day['amount']['value'] # Value is usually in USD
            print(f"{date_str:<15} | ${amount:.4f}")
            
    else:
        print(f"Error {response.status_code}: {response.text}")


# This is the main method
# if __name__ == "__main__":
#     get_daily_costs()


print(get_daily_costs())
```
