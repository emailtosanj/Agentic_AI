```python
import openai

# 1. Define current pricing (Price per 1M tokens)
# Update these values based on the February 2026 rates

#CAUTION:
#PRICING is not standard and varies as per usage and rate card applied 
# - from the GenAI LLM provider.
# Most of the models usage is compatible with OpenAI API, thus having the  ChatCompletion can be used below
# However, there could be variations in that case check the respective API dcoumentation which is used.

PRICING = {
    "gpt-5.2": {"input": 1.75, "output": 14.00},
    "gpt-5-mini": {"input": 0.25, "output": 2.00},
    "o3-mini": {"input": 1.10, "output": 4.40},
    "gpt-4o": {"input": 2.50, "output": 10.00},
}


def get_completion_with_cost(model_id, prompt):
    client = openai.OpenAI()
    
    # Make the API call
    response = client.chat.completions.create(
        model=model_id,
        messages=[{"role": "user", "content": prompt}]
    )

    # 2. Extract token usage
    usage = response.usage
    prompt_tokens = usage.prompt_tokens
    completion_tokens = usage.completion_tokens
    
    # 3. Calculate Cost
    prices = PRICING.get(model_id, {"input": 0, "output": 0})
    input_cost = (prompt_tokens / 1_000_000) * prices["input"]
    output_cost = (completion_tokens / 1_000_000) * prices["output"]
    total_cost = input_cost + output_cost

    return {
        "content": response.choices[0].message.content,
        "tokens": {"input": prompt_tokens, "output": completion_tokens},
        "cost_usd": f"${total_cost:.6f}"
    }

# Example Usage
result = get_completion_with_cost("gpt-5-mini", "Write a 50-word poem about AI.")
print(f"Response: {result['content']}")
print(f"Usage: {result['tokens']}")
print(f"Total Cost: {result['cost_usd']}")
```
