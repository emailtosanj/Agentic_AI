```python
from google import genai


# This automatically picks up the GEMINI_API_KEY environment variable
# client = genai.Client()

from google import genai
from google.genai import types, errors

client = genai.Client()

messages = [{"role": "user", "content": "Pick up a business area in a topi-tier banking insititution or financial industry company a business area which is worth exploring for an Agentic AI opportunity."}]


# Assuming your messages_sanj is already mapped to gemini_contents (from the previous step)

gemini_contents = [
    types.Content(
        role=m["role"], 
        parts=[types.Part.from_text(text=m["content"])]
    ) for m in messages
]


try:
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=gemini_contents,
        config=types.GenerateContentConfig(
            temperature=0.7,
        )
    )

    # 1. Check for Content/Safety Blocks
    # The API call succeeded, but the model might have stopped early due to safety filters or max tokens.
    if response.candidates and response.candidates[0].finish_reason != "STOP":
        print(f"Warning: Generation stopped early. Reason: {response.candidates[0].finish_reason}")
        # You can inspect response.candidates[0].safety_ratings here if needed
    
    # If all is good, print/return the text
    elif response.text:
        print("Success:")
        print(response.text)
    else:
        print("Error: Received an empty response.")

# 2. Catch Specific Google API Errors
except errors.APIError as e:
    # This catches rate limits (429), authentication issues (401), and server errors (500)
    print(f"Google API Error: {e.message}")
    print(f"Status Code: {e.code}")
    
    if e.code == 429:
        print("Action: Implement an exponential backoff retry here.")
    elif e.code == 401:
        print("Action: Check if your GEMINI_API_KEY is loaded correctly.")

# 3. Catch Everything Else (Network drops, malformed data, etc.)
except Exception as e:
    print(f"An unexpected system error occurred: {e}")
```
